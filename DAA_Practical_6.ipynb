```python
import time


# Matrix Chain Multiplication using Dynamic Programming
def matrix_chain_order(p):
    n = len(p) - 1

    # dp[i][j] stores the minimum multiplication cost
    # for matrices Ai to Aj
    dp = [[0 for _ in range(n)] for _ in range(n)]

    # length is the chain length
    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (
                    dp[i][k]
                    + dp[k + 1][j]
                    + p[i] * p[k + 1] * p[j + 1]
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[0][n - 1]


# User input
n = int(input("Enter the number of matrices: "))

if n <= 0:
    print("Number of matrices must be greater than 0.")
else:
    print("\nEnter the dimensions of the matrices.")

    dimensions = []

    rows = int(input("Enter rows of Matrix 1: "))
    columns = int(input("Enter columns of Matrix 1: "))

    dimensions.append(rows)
    dimensions.append(columns)

    previous_columns = columns

    for i in range(2, n + 1):
        rows = previous_columns
        columns = int(input(
            f"Enter columns of Matrix {i} "
            f"(rows must be {rows}): "
        ))

        dimensions.append(columns)
        previous_columns = columns

    # Measure execution time
    start_time = time.perf_counter()

    minimum_cost = matrix_chain_order(dimensions)

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    # Display results
    print("\n--- Matrix Chain Multiplication ---")

    for i in range(n):
        print(
            f"Matrix {i + 1}: "
            f"{dimensions[i]} x {dimensions[i + 1]}"
        )

    print("\nMinimum number of scalar multiplications:",
          minimum_cost)

    print("Execution Time:", execution_time, "seconds")
    print("Time Complexity: O(n^3)")
    print("Space Complexity: O(n^2)")
```
