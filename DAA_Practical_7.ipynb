```python
import time

def min_coins(coins, amount):
    # dp[i] = minimum number of coins needed to make amount i
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0

    # Build the DP table
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)

    return dp[amount]


# User input
coins = list(map(int, input("Enter coin denominations separated by spaces: ").split()))
amount = int(input("Enter the amount to make: "))

# Validate input
if amount < 0 or any(coin <= 0 for coin in coins):
    print("Please enter valid positive values.")
else:
    # Measure execution time
    start_time = time.perf_counter()

    result = min_coins(coins, amount)

    end_time = time.perf_counter()
    execution_time = end_time - start_time

    # Display result
    print("\n--- Making Change Using Dynamic Programming ---")

    if result == float('inf'):
        print("Change cannot be made with the given coins.")
    else:
        print("Coins:", coins)
        print("Amount:", amount)
        print("Minimum number of coins required:", result)

    print("Execution Time:", execution_time, "seconds")
    print("Time Complexity: O(n × A)")
    print("Space Complexity: O(A)")
```
